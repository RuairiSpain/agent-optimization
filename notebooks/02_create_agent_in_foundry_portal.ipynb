# 02 — Create the agent in Foundry Agent Service

**Foundry feature:** creating a **prompt agent** in Foundry Agent Service — role, model, and
function tools. **Mode: Portal.** There's no API for this step in the current preview, so this
notebook is a guided walkthrough rather than executable code; the code cells only print the exact
text you'll paste into the portal.

```python
import json, subprocess, sys
from pathlib import Path

# Repo layout: this notebook lives in notebooks/, the pack lives in ../prompt-agent-optimizer-baselines
PACK_ROOT = Path("..").resolve() / "prompt-agent-optimizer-baselines"
AGENT_ID = "01-travel-approval-strict"          # <- the one case study every notebook in this series uses
AGENT_DIR = PACK_ROOT / AGENT_ID

assert AGENT_DIR.exists(), f"Can't find {AGENT_DIR} -- run this notebook from a checkout of the repo."

def run(cmd, cwd=PACK_ROOT):
    """Run a pack CLI tool and print its output, the way you would from a terminal."""
    print("$", " ".join(cmd))
    result = subprocess.run(cmd, cwd=cwd, capture_output=True, text=True)
    print(result.stdout)
    if result.returncode != 0:
        print(result.stderr)
    return result

print(f"Pack root : {PACK_ROOT}")
print(f"Case study: {AGENT_ID}")
```

## Step-by-step

1. In the Foundry portal, choose **Create agent → Prompt agent**.
2. **Name** it `travel-approval-strict` (or any name — this is a label, not something the optimizer
   checks against `agent.yaml`).
3. **Model**: pick the model named in `agent.yaml` (`model: gpt-4.1-mini` for this case study, or the
   closest equivalent available in your Foundry resource).
4. **Instructions**: paste the full contents of `instructions.md`, printed below.
5. **Tools**: add the three function tools from `tools.json`, one at a time, using the printed JSON
   as the schema for each (name, description, parameters).
6. Save the agent, and confirm it responds to a test message in the portal's chat playground before
   moving on — that's your signal the instructions and tool schema parsed correctly.
